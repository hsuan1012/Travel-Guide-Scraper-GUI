# Cell 2：爬取「台灣旅遊」書籍資料
這段程式碼負責自動化抓取誠品書店「台灣旅遊」分類的書籍資訊。程式會啟動瀏覽器並透過迴圈自動翻頁，利用 CSS 選擇器解析網頁結構，提取出每一本書的「書名」與「作者」，最後利用 Pandas 將收集到的資料儲存成「台灣旅遊書推薦.csv」檔案。

```python
#誠品書局-旅遊-台灣旅遊
from selenium.webdriver.common.by import By  #誠品為動態網頁
import undetected_chromedriver as uc
import time
import pandas as pd

driver = uc.Chrome()

#儲存書名和作者
books = []
authors = []

#換頁爬取多頁
for page_num in range(0, 19):
    #根據頁數修改start參數
    url = f"https://www.eslite.com/category/3/128?final_price=0,&publishDate=0&sort=manufacturer_date+desc&size=40&display=square&start={page_num}&status=add_to_shopping_cart&categories=[%22128%22]&exp=o"
    driver.get(url)

    time.sleep(5)

    #抓取<span data-v-b5b956c0>內容
    elements = driver.find_elements(By.CSS_SELECTOR, "span[data-v-b5b956c0]")

    #書名和作者
    for idx, element in enumerate(elements, start=1):
        if idx % 2 == 1:
            books.append(element.text)  #奇數(書名)
        else:
            authors.append(element.text)  #偶數(作者)


df = pd.DataFrame({'書名': books, '作者': authors})


for book, author in zip(books, authors):
    print(f"書名: {book}")
    print(f"作者: {author}")

#儲存成CSV檔
df.to_csv("台灣旅遊書推薦.csv", index=False, encoding="utf-8")

driver.quit()
```

# Cell 7：建置旅遊書籍推薦系統 UI 介面
這段程式碼整合了前面爬取到的所有 CSV 資料，並利用 Tkinter 模組建立一個圖形化使用者介面（GUI）。使用者可以在介面上透過下拉式選單選擇想查詢的「區域」（如台灣、亞洲、美洲等），系統會連動更新對應的「國家/地點」選單。點擊搜尋後，程式會讀取對應的 CSV 檔案，過濾出符合該地點條件的書單，並將結果顯示在文字視窗中。

```python
#UI介面(旅遊書籍推薦系統)
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk
import pandas as pd


def search_books():
    region = selected_region.get().strip()  #選擇區域
    country = selected_country.get().strip()  #選擇地點
    if not country or not region:
        messagebox.showerror("錯誤", "請選擇或輸入區域和地點！")
        return

    try:
        # 根據選擇的區域讀取對應的CSV檔
        if region == "亞洲旅遊":
            df = pd.read_csv("亞洲旅遊書推薦.csv", encoding="utf-8")
        elif region == "美洲旅遊":
            df = pd.read_csv("美洲旅遊書推薦.csv", encoding="utf-8")
        elif region == "歐洲旅遊":
            df = pd.read_csv("歐洲旅遊書推薦.csv", encoding="utf-8")
        elif region == "台灣旅遊":
            df = pd.read_csv("台灣旅遊書推薦.csv", encoding="utf-8")
        elif region == "非洲旅遊":
            df = pd.read_csv("非洲旅遊書推薦.csv", encoding="utf-8")
        else:
            messagebox.showerror("錯誤", "未知區域選擇！")
            return

        #篩選包含國家名稱的書籍
        recommendations = df[df['書名'].str.contains(country, na=False)]

        result_text.delete(1.0, tk.END)

        if not recommendations.empty:
            for _, row in recommendations.iterrows():
                result_text.insert(
                    tk.END,
                    f"書名: {row['書名']}\n作者: {row['作者']}\n\n"
                )
        else:
            result_text.insert(tk.END, "找不到符合條件的書籍\n")
    except FileNotFoundError:
        messagebox.showerror("錯誤", "找不到選擇的區域CSV檔案")
    except Exception as e:
        messagebox.showerror("錯誤", f"發生錯誤：{e}")


def update_country_options(event):
    #根據選擇的區域顯示國家選項
    region = selected_region.get()
    if region == "台灣旅遊":
        countries = ["台北", "台中", "台南", "高雄", "花蓮"]
    elif region == "亞洲旅遊":
        countries = ["日本", "韓國", "泰國", "越南", "新加坡"]
    elif region == "美洲旅遊":
        countries = ["美國", "加拿大", "墨西哥", "巴西", "古巴"]
    elif region == "歐洲旅遊":
        countries = ["英國", "法國", "德國", "西班牙", "義大利"]
    elif region == "非洲旅遊":
        countries = ["埃及", "摩洛哥", "肯亞", "衣索比亞", "馬達加斯加"]
    else:
        countries = []

    selected_country['values'] = countries
    selected_country.set("")


#主視窗
root = tk.Tk()
root.title("旅遊書籍推薦系統")
root.geometry("900x500")

label_region = tk.Label(root, text="請選擇區域：", font=("Arial", 14))
label_region.pack(pady=10)

#下拉選單(區域選擇）
regions = ["台灣旅遊", "亞洲旅遊", "美洲旅遊", "歐洲旅遊", "非洲旅遊"]
selected_region = ttk.Combobox(root, values=regions, font=("Arial", 12))
selected_region.set("")
selected_region.pack(pady=10)

#當選擇區域時更新國家選項
selected_region.bind("<<ComboboxSelected>>", update_country_options)

label_country = tk.Label(root, text="請選擇或輸入地點：", font=("Arial", 14))
label_country.pack(pady=10)

#下拉選單和輸入框(國家選擇)
selected_country = ttk.Combobox(root, font=("Arial", 12))
selected_country.set("")
selected_country.pack(pady=10)

#搜尋按鈕
search_button = tk.Button(root, text="搜尋", font=("Arial", 14), command=search_books)
search_button.pack(pady=10)

#滾動條與文字框架
frame = tk.Frame(root)
frame.pack(pady=10, fill=tk.BOTH, expand=True)

#滾動條
scrollbar = tk.Scrollbar(frame)
scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

result_text = tk.Text(frame, font=("Arial", 12), wrap=tk.WORD, yscrollcommand=scrollbar.set)
result_text.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

scrollbar.config(command=result_text.yview)

root.mainloop()
```
